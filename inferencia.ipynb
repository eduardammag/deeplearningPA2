# Inferencia MOT17
Carregue uma sequencia MOT17, rode o baseline ou checkpoint temporal e exporte IDs consistentes.

```python
from pathlib import Path
from mot_pa2.mot17 import load_ground_truth, load_public_detections
from mot_pa2.evaluation import evaluate_baseline

sequence_root = Path('data/MOT17/train/MOT17-04')
detections = load_public_detections(sequence_root, detector='FRCNN')
ground_truth = load_ground_truth(sequence_root)
metrics, predictions, truth = evaluate_baseline(detections, ground_truth, matcher='hungarian')
metrics
```

Os objetos sao coloridos por `track_id` usando `mot_pa2.visualization.draw_frame`; o numero de identidades unicas e `len({item['track_id'] for frame in predictions for item in frame})`.
